```python
#Function for Even/Odd
def check_even_odd(number):

    if number % 2 == 0:
        return "Even"
    else:
        return "odd"


result = check_even_odd(25)

print(result)
```

```python
#Function for Large of Two Numbers
def largest(a, b):

    if a > b:
        return a
    else:
        return b

result = largest(50, 30)
print("Largest:", result)
```

```python
#Function + User Input
def square(number):
    return number * number


number = int(input("Enter number: "))

result = square(number)

print("Square:", result)
```

```python
#Function + Loop
def multiplication_table(number):

    for i in range(1, 11):
        print(number, "x", i, "=", number * i)


number = int(input("Enter number: "))
multiplication_table(number)
```

```python
#Largest of Three Numbers
def largest(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >=c:
        return b
    else:
        return c

a = int(input("Enter a: " ))
b = int(input("Enter b: " ))
c = int(input("Enter c: " ))

print("Largest =", largest(a, b, c))
```

```python
#Grade calculation
def find_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "c"
    elif marks >= 60:
        return "D"
    elif marks >= 40:
        return "E"
    else:
        return "Fail"
marks = int(input("Enter marks: "))

print("Grade =", find_grade(marks))
```

```python
#sum of 1 to N
def sum_n(num):
    total = 0 

    for i in range(1, num +1):
        total += i

    return total

n = int(input("Enter n: "))

print("sum =", sum_n(n))
```
